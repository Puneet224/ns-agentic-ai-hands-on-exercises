# Step 2: Safe File I/O & JSON Persistence

This notebook demonstrates how to reuse the agent module from the src package and save configuration data as JSON.

```python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
src_path = project_root.parent / 'src'
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from agent_core import create_default_profile

profile = create_default_profile()
print(profile.to_json())

config_path = Path('agent_config.json')
saved_path = profile.save_to_json(config_path)
print(f'Configuration securely saved to {saved_path}.')
```
